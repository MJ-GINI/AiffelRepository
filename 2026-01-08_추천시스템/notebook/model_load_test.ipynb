```python
import os
import sys
import numpy as np
import tensorflow as tf

# =========================================================
# 1. 경로 설정 (노트북 위치 문제 해결)
# =========================================================

# 현재 노트북이 있는 폴더 (aiffel/autoint/notebook)
current_notebook_dir = os.getcwd()

# 한 단계 위 폴더 (aiffel/autoint) -> 여기가 진짜 프로젝트 루트입니다.
project_root_dir = os.path.dirname(current_notebook_dir)

# 파이썬에게 "위쪽 폴더도 검색 경로에 추가해줘!" 라고 명령
if project_root_dir not in sys.path:
    sys.path.append(project_root_dir)

print(f"📂 현재 노트북 위치: {current_notebook_dir}")
print(f"📂 프로젝트 루트 위치: {project_root_dir}")

# =========================================================
# 2. 파일 및 모듈 로드
# =========================================================

# 1) 모듈 불러오기 (이제 경로를 추가했으니 불러와질 겁니다)
try:
    from autoint_mlp import AutoIntMLPModel
    print("✅ 'autoint_mlp.py' 모듈 로드 성공!")
except ImportError:
    print("❌ 오류: 'autoint_mlp.py'를 여전히 못 찾고 있습니다. 파일명이 정확한지 확인해주세요.")

# 2) 파일 경로 재설정 (프로젝트 루트 기준)
# 데이터는 aiffel/autoint/data 에 있다고 가정합니다.
data_path = os.path.join(project_root_dir, 'data') 
model_path = os.path.join(project_root_dir, 'model')
weights_file = os.path.join(model_path, 'autoInt_model_weights.weights.h5')
field_dims_file = os.path.join(data_path, 'field_dims.npy')

# 3) 가중치 파일 존재 확인
if os.path.exists(weights_file):
    print(f"✅ 가중치 파일 발견: {weights_file}")
else:
    print(f"❌ 가중치 파일 없음: {weights_file}")
    # 혹시 파일명이 다를 수 있으니 model 폴더 내용물 출력
    if os.path.exists(model_path):
        print(f"   (참고) model 폴더 내 파일들: {os.listdir(model_path)}")

# 4) 데이터 차원 로드
try:
    field_dims = np.load(field_dims_file)
    print(f"✅ field_dims 로드 성공 (크기: {len(field_dims)})")
except:
    print(f"⚠️ field_dims 로드 실패. 경로 확인 필요: {field_dims_file}")
    # 테스트를 위한 임시 값 (실패 시에도 코드 돌아가게 함)
    field_dims = np.array([6040, 3706, 10, 81, 4, 12, 1, 18, 18, 16, 2, 7, 21, 3439], dtype=np.int64)

# =========================================================
# 3. 모델 빌드 및 가중치 로드 테스트
# =========================================================

# 모델 파라미터 (학습 코드와 동일해야 함)
embed_dim = 16
dropout = 0.4
dnn_hidden_units = (32, 32)
dnn_activation = 'relu'

print("\n🛠️ 모델 빌드 중...")
try:
    tf.keras.backend.clear_session()
    
    model = AutoIntMLPModel(
        field_dims=field_dims,
        embedding_size=embed_dim,
        att_layer_num=3,
        att_head_num=2,
        att_res=True,
        dnn_hidden_units=dnn_hidden_units,
        dnn_activation=dnn_activation,
        l2_reg_dnn=0,
        l2_reg_embedding=1e-5,
        dnn_use_bn=False,
        dnn_dropout=dropout,
        init_std=0.0001
    )
    
    # 더미 데이터로 초기화
    dummy_input = tf.zeros((1, len(field_dims)), dtype=tf.int64)
    model(dummy_input)
    print("✅ 모델 초기화(Build) 성공!")
    
except Exception as e:
    print(f"❌ 모델 빌드 실패: {e}")

print("\n⚖️ 가중치 로드 시도...")
try:
    model.load_weights(weights_file)
    print("\n🎉🎉🎉 [최종 성공] 가중치가 완벽하게 로드되었습니다! 🎉🎉🎉")
    print("이제 이 모델 파일(.h5)은 정상임이 확인되었습니다.")
    
except Exception as e:
    print("\n❌❌❌ [실패] 가중치 로드 중 오류 발생 ❌❌❌")
    print(e)
```
